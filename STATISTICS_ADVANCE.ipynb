**2. What is a p-value, and how is it used in hypothesis testing? What does it mean if the p-value is very small (e.g., 0.01)?**

### **P-Value in Hypothesis Testing**  
A **p-value** is the probability of observing data as extreme as, or more extreme than, the current results, assuming the **null hypothesis (\(H_0\)) is true**.  

### **How It's Used:**  
- **Compare p-value to significance level (\(\alpha\))**:  
  - If \( p \leq \alpha \) → Reject \(H_0\) (significant result).  
  - If \( p > \alpha \) → Fail to reject \(H_0\) (not significant).  

### **If P-Value is Very Small (e.g., 0.01)**  
- **Strong evidence against \(H_0\)**  
- **Highly unlikely due to chance**  
- **Reject \(H_0\) at 1% significance level**  

Example: If a drug test gives **p = 0.01**, it's likely effective since the result is **statistically significant** at 95% confidence.

```python
import scipy.stats as stats

# Given data
sample_mean = 105
population_mean = 100
population_std = 15
sample_size = 30

# Calculate z-score
z_score = (sample_mean - population_mean) / (population_std / (sample_size ** 0.5))

# Get p-value (for a one-tailed test)
p_value = 1 - stats.norm.cdf(z_score)

print(f"Z-Score: {z_score:.2f}")
print(f"P-Value: {p_value:.4f}")

# Decision
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis (significant result).")
else:
    print("Fail to reject the null hypothesis (not significant).")
```

**3.Compare and contrast the binomial and Bernoulli distributions**

The **Bernoulli distribution** models a single trial with two possible outcomes: success (\(X = 1\)) with probability \( p \) and failure (\(X = 0\)) with probability \( 1 - p \). In contrast, the **binomial distribution** extends this by modeling the number of successes in \( n \) independent Bernoulli trials, where each trial has the same probability \( p \) of success. The Bernoulli distribution is a special case of the binomial distribution when \( n = 1 \). The mean and variance of a Bernoulli distribution are \( p \) and \( p(1 - p) \), respectively, while for a binomial distribution, they are \( np \) and \( np(1 - p) \), showing that the binomial scales with the number of trials. While the Bernoulli distribution is used for single yes/no experiments, the binomial distribution is useful for counting successes over multiple trials, such as the number of heads in multiple coin flips.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom

# Parameters
p = 0.5  # Probability of success
n = 10    # Number of trials for Binomial

# Bernoulli distribution (Single trial)
bernoulli_samples = bernoulli.rvs(p, size=1000)

# Binomial distribution (Multiple trials)
binomial_samples = binom.rvs(n, p, size=1000)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(bernoulli_samples, bins=[-0.5, 0.5, 1.5], edgecolor='black', alpha=0.7)
plt.xticks([0, 1])
plt.title("Bernoulli Distribution (p=0.5)")
plt.xlabel("Outcome (0=Failure, 1=Success)")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(binomial_samples, bins=np.arange(0, n+2)-0.5, edgecolor='black', alpha=0.7)
plt.xticks(range(n+1))
plt.title(f"Binomial Distribution (n={n}, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

**4. Under what conditions is the binomial distribution used, and how does it relate to the Bernoulli distribution?**

### **Conditions for Using the Binomial Distribution**  
The **binomial distribution** is used when the following conditions are met:  

1. **Fixed Number of Trials (\( n \))** – The experiment consists of a set number of independent trials.  
2. **Binary Outcomes** – Each trial has only two possible outcomes: success (\(X = 1\)) or failure (\(X = 0\)).  
3. **Constant Probability (\( p \))** – The probability of success remains the same for all trials.  
4. **Independent Trials** – The outcome of one trial does not affect the others.  

### **Relationship to the Bernoulli Distribution**  
The **binomial distribution** is essentially a **sum of multiple independent Bernoulli trials**. A **Bernoulli distribution** models a **single** trial with two possible outcomes (success or failure), while the **binomial distribution** generalizes this to **\( n \) repeated Bernoulli trials**, counting the number of successes. Mathematically, if \(X_1, X_2, ..., X_n\) are independent Bernoulli(\( p \)) random variables, then the binomial random variable \( X \) follows:  

\[
X = X_1 + X_2 + ... + X_n \sim \text{Binomial}(n, p)
\]

Thus, the **Bernoulli distribution** is a special case of the **binomial distribution** where \( n = 1 \).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom

# Parameters
p = 0.5  # Probability of success
n = 10   # Number of trials

# Simulate 10 independent Bernoulli trials
bernoulli_trials = bernoulli.rvs(p, size=(1000, n))  # 1000 simulations of 10 trials
binomial_counts = np.sum(bernoulli_trials, axis=1)  # Sum of successes in each simulation

# Generate binomial samples directly
binomial_samples = binom.rvs(n, p, size=1000)

# Plot results
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(np.sum(bernoulli_trials, axis=1), bins=np.arange(0, n+2)-0.5, edgecolor='black', alpha=0.7)
plt.xticks(range(n+1))
plt.title("Sum of 10 Bernoulli Trials (1000 Simulations)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(binomial_samples, bins=np.arange(0, n+2)-0.5, edgecolor='black', alpha=0.7)
plt.xticks(range(n+1))
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

**5.What are the key properties of the Poisson distribution, and when is it appropriate to use this distribution?**

### **Key Properties of the Poisson Distribution**  
The **Poisson distribution** models the probability of a given number of events occurring in a fixed interval of time or space, under the assumption that these events happen at a constant rate and independently of each other. It is defined by a single parameter \( \lambda \), which represents both the **mean** and **variance** of the distribution.  

1. **Probability Mass Function (PMF):**  
   \[
   P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}, \quad k = 0, 1, 2, \dots
   \]  
   where \( \lambda \) is the average number of events per interval.  

2. **Mean and Variance:**  
   - **Mean**: \( E[X] = \lambda \)  
   - **Variance**: \( \text{Var}(X) = \lambda \)  

3. **Memoryless Property (Approximate):**  
   - Unlike the exponential distribution, the Poisson distribution does **not** strictly have the memoryless property, but it can approximate independent arrivals over time.  

4. **Additivity:**  
   - If \( X_1 \sim \text{Poisson}(\lambda_1) \) and \( X_2 \sim \text{Poisson}(\lambda_2) \), then their sum \( X_1 + X_2 \sim \text{Poisson}(\lambda_1 + \lambda_2) \).  

### **When to Use the Poisson Distribution**  
The **Poisson distribution** is appropriate when:  
1. **Events occur randomly and independently** over a fixed interval of time or space.  
2. **The average rate (\( \lambda \)) of occurrence is constant**, meaning events do not cluster.  
3. **Only one event can happen at a time**, meaning no simultaneous occurrences.  
4. **The number of trials is very large and the probability of success in each trial is very small**, making it a good approximation of the binomial distribution when \( n \) is large and \( p \) is small (\( np \approx \lambda \)).

### **Examples of Poisson Distribution Use Cases:**  
- The number of customer arrivals at a store per hour.  
- The number of emails received in a day.

 A **Probability Mass Function (PMF)** is used for **discrete** random variables and gives the probability of each specific outcome. For example, in a fair die roll, the probability of rolling a 4 is \( P(X = 4) = 1/6 \). In contrast, a **Probability Density Function (PDF)** applies to **continuous** random variables and represents the probability density at a given point, but the probability of any exact value is zero. Instead, probabilities are found by integrating over a range. For instance, in a normal distribution of human heights, the probability of being exactly **170 cm** tall is zero, but the probability of being between **170 cm and 175 cm** is given by integrating the PDF over that interval. The key difference is that **PMF gives exact probabilities for discrete values**, while **PDF provides a density function for continuous values, requiring integration to find probabilities over an interval**.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm

# PMF: Binomial Distribution (Discrete)
n, p = 10, 0.5  # 10 trials, 50% success probability
x_pmf = np.arange(0, n+1)  # Possible values (0 to 10)
y_pmf = binom.pmf(x_pmf, n, p)  # Compute PMF

# PDF: Normal Distribution (Continuous)
mu, sigma = 0, 1  # Mean=0, Std Dev=1
x_pdf = np.linspace(-4, 4, 1000)  # Continuous values
y_pdf = norm.pdf(x_pdf, mu, sigma)  # Compute PDF

# Plot PMF and PDF
plt.figure(figsize=(12, 5))

# Plot PMF (Discrete)
plt.subplot(1, 2, 1)
# Remove the `use_line_collection` argument as it's not supported in older versions
plt.stem(x_pmf, y_pmf)
plt.title("PMF - Binomial(n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.xticks(x_pmf)

# Plot PDF (Continuous)
plt.subplot(1, 2, 2)
plt.plot(x_pdf, y_pdf, color='red')
plt.title("PDF - Normal(μ=0, σ=1)")
plt.xlabel("Value")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

**7.Explain the Central Limit Theorem (CLT) with example.**

The Central Limit Theorem (CLT) states that, regardless of the original population distribution, the sampling distribution of the sample mean will approximate a normal distribution as the sample size
𝑛
n increases, provided
𝑛
n is sufficiently large (typically
𝑛
≥
30
n≥30).



```python
import numpy as np
import matplotlib.pyplot as plt

# Original population: Exponential distribution (highly skewed)
population = np.random.exponential(scale=2, size=10000)

# Function to compute sample means
def sample_means(sample_size, num_samples):
    means = [np.mean(np.random.choice(population, sample_size)) for _ in range(num_samples)]
    return means

# Sample sizes
sample_sizes = [5, 30, 100]

plt.figure(figsize=(12, 5))
for i, n in enumerate(sample_sizes):
    plt.subplot(1, 3, i+1)
    means = sample_means(n, 1000)  # 1000 sample means
    plt.hist(means, bins=30, density=True, alpha=0.7, color='blue')
    plt.title(f"Sample Size = {n}")

plt.suptitle("Central Limit Theorem: Sampling Distribution of the Mean")
plt.show()
```

**8. Compare z-scores and t-scores. When should you use a z-score, and when should a t-score be applied instead?**

A z-score and a t-score both measure how far a data point is from the mean in terms of standard deviations, but they are used in different situations. A z-score is used when the population standard deviation is known and the sample size is large, typically \( n \geq 30 \), and it follows a standard normal distribution. In contrast, a t-score is used when the population standard deviation is unknown and the sample size is small, typically \( n < 30 \), and it follows a t-distribution, which has heavier tails to account for extra variability in small samples. As the sample size increases, the t-distribution approaches the normal distribution, making the difference between z-scores and t-scores less significant for larger samples.

```python
import scipy.stats as stats

# Given data
x = 75       # Sample mean
mu = 70      # Population mean
sigma = 10   # Population std dev (known)
s = 12       # Sample std dev (unknown sigma)
n_large = 50 # Large sample size (use Z-score)
n_small = 15 # Small sample size (use T-score)

# Z-score (when sigma is known and sample is large)
z_score = (x - mu) / (sigma / (n_large ** 0.5))
print(f"Z-score (n={n_large}):", z_score)

# T-score (when sigma is unknown and sample is small)
t_score = (x - mu) / (s / (n_small ** 0.5))
t_value = stats.t.ppf(0.975, df=n_small-1)  # Critical t-value for 95% confidence
print(f"T-score (n={n_small}):", t_score, "Critical t-value:", t_value)
```

**9.Given a sample mean of 105, a population mean of 100, a standard deviation of 15, and a sample size of 25, calculate the z-score and p-value. Based on a significance level of 0.05, do you reject or fail to reject the null hypothesis?Task: Write Python code to calculate the z-score and p-value for the given data.Objective: Apply the formula for the z-score and interpret the p-value for hypothesis testing.**

```python
import scipy.stats as stats

# Given data
sample_mean = 72
population_mean = 70
std_dev = 8
sample_size = 40
alpha = 0.01  # Significance level

# Calculate z-score
z_score = (sample_mean - population_mean) / (std_dev / (sample_size ** 0.5))

# Calculate p-value (two-tailed test)
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

# Decision based on p-value
decision = "Reject the null hypothesis" if p_value < alpha else "Fail to reject the null hypothesis"

# Output results
print(f"Z-score: {z_score:.2f}")
print(f"P-value: {p_value:.4f}")
print(f"Decision: {decision}")
```

**10.Simulate a binomial distribution with 10 trials and a probability of success of 0.6 using Python.Generate 1,000 samples and plot the distribution. What is the expected mean and variance?Task: Use Python to generate the data, plot the distribution, and calculate the mean and variance.Objective: Understand the properties of a binomial distribution and verify them through simulation.**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import binom

# Parameters for binomial distribution
n = 10      # Number of trials
p = 0.6     # Probability of success
size = 1000 # Number of samples

# Generate binomial samples
samples = np.random.binomial(n, p, size)

# Calculate expected mean and variance
expected_mean = n * p
expected_variance = n * p * (1 - p)

# Calculate sample mean and variance
sample_mean = np.mean(samples)
sample_variance = np.var(samples, ddof=1)  # Using ddof=1 for sample variance

# Plot the distribution
plt.figure(figsize=(10, 5))
sns.histplot(samples, bins=np.arange(n+2)-0.5, kde=False, stat="density", color="blue", edgecolor="black")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution (n=10, p=0.6, 1000 samples)")
plt.xticks(range(n+1))
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Display results
print(f"Expected Mean: {expected_mean}")
print(f"Expected Variance: {expected_variance}")
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Variance: {sample_variance:.2f}")

plt.show()
```

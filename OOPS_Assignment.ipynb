**3.Explain the difference between instance methods and class methods. Provide an example of each?**

Instance methods are functions defined inside a class that operate on instances (objects) of the class. These methods can access and modify the object's attributes and perform operations specific to each instance.They take self as their first parameter, which refers to the current instance of the class.

```python
#example of instance method

class Restaurant:
    def __init__(self, name, location, menu):
        self.name = name
        self.location = location
        self.menu = menu

    def get_details(self):
        # This method uses instance-specific data
        return f"Restaurant: {self.name}, Location: {self.location}, Menu: {', '.join(self.menu)}"

# Creating an instance of Restaurant
my_restaurant = Restaurant("Bella's Bistro", "New York", ["Pasta", "Pizza", "Salad"])

# Calling the instance method
print(my_restaurant.get_details())  # Output: Restaurant: Bella's Bistro, Location: New York, Menu: Pasta, Pizza, Salad
```

Class methods are functions that are bound to the class rather than instances. They operate on the class itself and not on individual objects. They can access and modify class-level attributes but not instance-specific data. They take cls as their first parameter, which refers to the class itself, not an instance

**4.How does Python implement method overloading? Give an example.?**

```python
#example of class method

class Restaurant:
    total_restaurants = 0  # Class-level attribute

    def __init__(self, name, location, menu):
        self.name = name
        self.location = location
        self.menu = menu
        Restaurant.total_restaurants += 1  # Increment the count whenever a new restaurant is created

    @classmethod
    def get_total_restaurants(cls):
        # This method operates on the class-level data
        return f"Total restaurants in business: {cls.total_restaurants}"

# Creating instances of Restaurant
restaurant1 = Restaurant("Bella's Bistro", "New York", ["Pasta", "Pizza", "Salad"])
restaurant2 = Restaurant("Joe's Diner", "Los Angeles", ["Burgers", "Fries"])

# Calling the class method
print(Restaurant.get_total_restaurants())  # Output: Total restaurants in business: 2
```

In Python, method overloading (the ability to define multiple methods with the same name but different arguments) does not exist in the same way as in some other languages (like Java or C++). In Python, a method can be redefined in a class to accept any number of arguments, and the behavior is typically managed using default arguments or variable-length arguments.

However, you can achieve a similar effect to method overloading by:

Using default arguments: You can provide default values for arguments in a method so it can be called with different numbers of arguments.
Using *args (for variable-length positional arguments) or **kwargs (for variable-length keyword arguments): This allows a method to accept any number of arguments, thus handling multiple use cases.

```python
#method overloading

class Calculator:
    def add(self, *args):
        # This method can handle any number of arguments
        return sum(args)

# Create an instance of the Calculator class
calc = Calculator()

# Call the method with different numbers of arguments
print(calc.add(2, 3))  # Output: 5 (2 + 3)
print(calc.add(1, 2, 3, 4, 5))  # Output: 15 (1 + 2 + 3 + 4 + 5)
```

```python
class Printer:
    def print_message(self, message="Hello", times=1):
        for _ in range(times):
            print(message)

# Create an instance of Printer
printer = Printer()

# Calling the method with different arguments
printer.print_message("Hello, World!", 3)  # Output: Prints "Hello, World!" three times
printer.print_message("Goodbye")           # Output: Prints "Goodbye" once (default times=1)
```

**6.Describe the five types of inheritance in Python. Provide a simple example of multiple inheritance?**

```python
#1. Single Inheritance
#Single inheritance involves a single base class and a single derived class. The derived class inherits from one parent class.

class Parent:
    def display(self):
        print("This is the Parent class")

class Child(Parent):
    def show(self):
        print("This is the Child class")

# Usage
child = Child()
child.display()
child.show()
```

```python
#multiple inheritance:Multiple inheritance occurs when a child class inherits from more than one parent class

class Father:
    def traits_father(self):
        print("Traits of father")

class Mother:
    def traits_mother(self):
        print("Traits of mother")

class Child(Father, Mother):
    def traits_child(self):
        print("Traits of child")

child = Child()
child.traits_father()  # Inherited from Father
child.traits_mother()  # Inherited from Mother
child.traits_child()   # Defined in Child
```

```python
#multilevel inheritance: Multilevel inheritance occurs when a derived class inherits from a base class, and another class further inherits from that derived class, forming a chain of inheritance.


class Grandparent:
    def traits_grandparent(self):
        print("Traits of Grandparent")

class Parent(Grandparent):
    def traits_parent(self):
        print("Traits of Parent")

class Child(Parent):
    def traits_child(self):
        print("Traits of Child")

child = Child()
child.traits_grandparent()  # Inherited from Grandparent
child.traits_parent()       # Inherited from Parent
child.traits_child()        # Defined in Child
```

```python
#hirerachial inheritance:Hierarchical inheritance happens when multiple child classes inherit from the same base class.

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

dog = Dog()
cat = Cat()

dog.speak()  # Inherited from Animal
dog.bark()   # Defined in Dog

cat.speak()  # Inherited from Animal
cat.meow()   # Defined in Cat
```

```python
#hybrid inheritance:Hybrid inheritance is a combination of two or more types of inheritance, typically a mix of hierarchical and multiple inheritance.


class Grandparent:
    def traits_grandparent(self):
        print("Traits of Grandparent")

class Parent(Grandparent):
    def traits_parent(self):
        print("Traits of Parent")

class Mother:
    def traits_mother(self):
        print("Traits of Mother")

class Child(Parent, Mother):
    def traits_child(self):
        print("Traits of Child")

child = Child()
child.traits_grandparent()  # Inherited from Grandparent
child.traits_parent()       # Inherited from Parent
child.traits_mother()       # Inherited from Mother
child.traits_child()        # Defined in Child
```

**7.What is the Method Resolution Order (MRO) in Python? How can you retrieve it programmatically??**

When Python encounters a method call, it uses the MRO to search the class hierarchy from top to bottom (starting from the current class and moving up to parent classes), and checks each class in the order defined by the MRO to find the method.

The order is determined by the C3 Linearization Algorithm, which Python uses to ensure that classes are considered in a specific order in the case of multiple inheritance. The primary goal of the MRO is to prevent the diamond problem (i.e., ambiguity in inheritance) by specifying a consistent search order.

```python
class A:
    def hello(self):
        print("Hello from A")

class B(A):
    def hello(self):
        print("Hello from B")

class C(A):
    def hello(self):
        print("Hello from C")

class D(B, C):
    pass

d = D()
d.hello()
```

**8Create an abstract base class Shape with an abstract method area(). Then create two subclasses Circle and Rectangle that implement the area() method.**

```python
from abc import ABC, abstractmethod
import math

# Abstract Base Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Subclass Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        # Area of a circle: π * radius^2
        return math.pi * (self.radius ** 2)

# Subclass Rectangle
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        # Area of a rectangle: width * height
        return self.width * self.height

# Create instances of Circle and Rectangle
circle = Circle(5)
rectangle = Rectangle(4, 6)

# Display areas
print(f"Area of Circle: {circle.area():.2f}")
print(f"Area of Rectangle: {rectangle.area()}")
```

**9.Demonstrate polymorphism by creating a function that can work with different shape objects to calculate and print their areas.**

Polymorphism allows us to use a single function to operate on different types of objects, provided they share the same interface (in this case, they implement the area() method). Since both the Circle and Rectangle classes share a common method area() (inherited from the abstract Shape class), we can create a function that works with any shape object to calculate and print the area.

```python
from abc import ABC, abstractmethod
import math

# Abstract Base Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Subclass Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        # Area of a circle: π * radius^2
        return math.pi * (self.radius ** 2)

# Subclass Rectangle
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        # Area of a rectangle: width * height
        return self.width * self.height

# Polymorphic function that works with any Shape object
def print_area(shape: Shape):
    print(f"The area of the shape is: {shape.area():.2f}")

# Create instances of Circle and Rectangle
circle = Circle(5)
rectangle = Rectangle(4, 6)

# Use the polymorphic function to calculate and print the area of different shapes
print_area(circle)      # This will calculate the area of the Circle
print_area(rectangle)   # This will calculate the area of the Rectangle
```

**10.Implement encapsulation in a BankAccount class with private attributes for balance and account_number. Include methods for deposit, withdrawal, and balance inquiry.**

```python
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.__account_number = account_number  # Private attribute
        self.__balance = initial_balance        # Private attribute

    # Method to deposit money into the account
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Deposit amount must be positive.")

    # Method to withdraw money from the account
    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"Withdrew ${amount}. New balance: ${self.__balance}")
            else:
                print("Insufficient funds for this withdrawal.")
        else:
            print("Withdrawal amount must be positive.")

    # Method to check the current balance
    def get_balance(self):
        return self.__balance

    # Method to get the account number (not a common practice to expose directly, but for demonstration)
    def get_account_number(self):
        return self.__account_number

# Create a BankAccount instance
account = BankAccount("123456789", 1000)

# Deposit money
account.deposit(500)

# Withdraw money
account.withdraw(200)

# Check balance
print(f"Current balance: ${account.get_balance()}")

# Accessing the account number
print(f"Account Number: {account.get_account_number()}")
```

**11.Write a class that overrides the __str__ and __add__ magic methods. What will these methods allow you to do?**

In Python, the magic methods (also known as dunder methods, short for "double underscore") allow you to define how instances of a class behave in certain situations. Two of these methods are:

__str__: This method is used to define the string representation of an object, which is returned when you use the str() function or when you print an object. It allows you to control how your object is represented as a string.

__add__: This method is used to define how objects of a class behave when the + operator is used with them. It allows you to define custom behavior for object addition.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Override the __str__ method to provide a custom string representation of the object
    def __str__(self):
        return f"Point({self.x}, {self.y})"

    # Override the __add__ method to define how two Point objects are added
    def __add__(self, other):
        if isinstance(other, Point):
            # Add the corresponding x and y values
            return Point(self.x + other.x, self.y + other.y)
        return NotImplemented

# Create two Point objects
point1 = Point(3, 4)
point2 = Point(1, 2)

# Printing a Point object uses the __str__ method
print(point1)  # Output: Point(3, 4)

# Adding two Point objects uses the __add__ method
point3 = point1 + point2
print(point3)  # Output: Point(4, 6)
```

**12.Explain the concept of the Diamond Problem in multiple inheritance. How does Python resolve it?**

The Diamond Problem is a well-known issue in object-oriented programming (OOP) that arises when a class inherits from multiple classes that have a common ancestor. This can lead to ambiguity when trying to determine which method or property of the ancestor class should be used in the derived class.Python resolves the Diamond Problem using a method called Method Resolution Order (MRO). The MRO defines the order in which Python will search for methods or attributes in a class hierarchy.     


```python
class A:
    def show(self):
        print("Method in class A")

class B(A):
    def show(self):
        print("Method in class B")

class C(A):
    def show(self):
        print("Method in class C")

class D(B, C):
    pass

d = D()
d.show()
```

**13.Create a decorator that measures and prints the execution time of a function.**

```python
import time

def measure_execution_time(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()  # Record the start time
        result = func(*args, **kwargs)  # Call the original function
        end_time = time.time()  # Record the end time
        execution_time = end_time - start_time  # Calculate the time difference
        print(f"Execution time of {func.__name__}: {execution_time:.6f} seconds")
        return result
    return wrapper

# Example usage of the decorator
@measure_execution_time
def example_function():
    # Simulating a time-consuming task
    time.sleep(2)

example_function()
```

**15.Implement a static method in a class that checks if a given year is a leap year.**

```python
class YearUtils:
    @staticmethod
    def is_leap_year(year):
        # Check if the year is divisible by 4, and not divisible by 100 unless also divisible by 400
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            return True
        return False

# Example usage:
print(YearUtils.is_leap_year(2020))  # Output: True
print(YearUtils.is_leap_year(2024))  # Output: True
print(YearUtils.is_leap_year(1900))  # Output: False
print(YearUtils.is_leap_year(2000))  # Output: True
```

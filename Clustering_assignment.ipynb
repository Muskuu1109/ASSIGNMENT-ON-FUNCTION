*Theoretical Questions*

**1. What is unsupervised learning in the context of machine learning?**

Unsupervised learning is a type of machine learning where the model is trained on data without labeled outcomes. Instead of learning from explicit instructions, it identifies patterns, relationships, or clusters in the data autonomously. Clustering algorithms like K-Means, DBSCAN, and hierarchical clustering fall under this category since they group data points based on similarities without predefined labels.


**2. What is clustering in machine learning?**  
Clustering is a technique in unsupervised learning where similar data points are grouped based on certain characteristics. It's used to identify natural patterns or structures in data, helping in customer segmentation, anomaly detection, and more.

**3. What is K-Means clustering and how does it work?**  
K-Means clustering is a popular algorithm that partitions data into **K** clusters. It follows these steps:
1. Choose **K** initial centroids randomly.
2. Assign each data point to the nearest centroid.
3. Compute new centroids as the average of assigned points.
4. Repeat steps 2-3 until centroids stabilize.

**4. How do you choose the optimal number of clusters in K-Means?**  
The **Elbow Method** is commonly used. It plots the sum of squared distances from each point to its cluster centroid. The point where the distortion curve bends (like an elbow) suggests the best **K**.

**5. What are the limitations of K-Means clustering?**  
- Sensitive to initial centroid placement.
- Assumes clusters are spherical and equally sized.
- Struggles with non-linear or overlapping clusters.
- Doesn't handle noise/outliers well.

**6. What is hierarchical clustering?**  
Hierarchical clustering builds a tree-like structure (**dendrogram**) of clusters.  
Two types:
- **Agglomerative:** Starts with individual points, merges clusters iteratively.
- **Divisive:** Starts with all data in one cluster, splits iteratively.

**7. What is DBSCAN and how does it differ from K-Means?**  
DBSCAN (**Density-Based Spatial Clustering of Applications with Noise**) groups dense regions of points and ignores outliers.  
Differences from K-Means:
- No need to specify **K**.
- Detects arbitrarily shaped clusters.
- Handles noise well.

**8. How does DBSCAN determine clusters?**  
DBSCAN relies on:
- **Epsilon (ε):** Defines the neighborhood radius.
- **MinPts:** Minimum points required in a neighborhood to form a dense cluster.

**9. What are the advantages of DBSCAN?**  
- Identifies complex-shaped clusters.
- Handles noise effectively.
- No need to predefine **K**.

**10. What are the limitations of DBSCAN?**  
- Sensitive to **ε** and **MinPts** selection.
- Struggles with varying-density clusters.

**11. What is the difference between hard and soft clustering?**  
- **Hard clustering:** Assigns each point to exactly one cluster (e.g., K-Means).
- **Soft clustering:** Allows a point to belong to multiple clusters with probabilities (e.g., Gaussian Mixture Models).

**12. What is the role of distance metrics in clustering?**  
Distance metrics (e.g., Euclidean, Manhattan, Cosine) determine similarity between points. The choice of metric affects clustering quality.

**13. How does feature scaling impact clustering algorithms?**  
Scaling ensures all features contribute equally.  
- K-Means and DBSCAN are sensitive to scale.
- Standardization (**z-score**) or normalization (**min-max**) is often used.

**14. What is mean shift clustering?**  
Mean shift is a density-based clustering method that moves points toward the densest region. It finds cluster centers by iteratively shifting centroids.

**15. What is spectral clustering?**  
Spectral clustering uses graph theory to group points. It transforms data into a similarity graph, then applies clustering (e.g., K-Means).

**16. How does clustering handle categorical data?**  
Some clustering methods work with categorical data:
- **K-Modes:** Extends K-Means for categorical features.
- **Hierarchical clustering:** Works with categorical similarity measures.

**17. How can clustering be used for anomaly detection?**  
Anomalies (outliers) are points that don’t fit well into clusters. Algorithms like DBSCAN can identify outliers as noise.

**18. What are real-world applications of clustering?**  
- Customer segmentation.
- Fraud detection.
- Image compression.
- Social network analysis.

**19. Can clustering be combined with supervised learning?**  
Yes! Clustering can be used for feature engineering or pre-processing before applying supervised learning models.

**20. What is the difference between partitioning-based, hierarchical, and density-based clustering?**  
- **Partitioning-based:** Divides data into **K** clusters (e.g., K-Means).
- **Hierarchical:** Builds a nested structure (e.g., Agglomerative clustering).
- **Density-based:** Groups dense areas and ignores noise (e.g., DBSCAN).

21. Generate synthetic data with 4 centers using make_blobs and apply K-Means



```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=4, random_state=42)
kmeans = KMeans(n_clusters=4, random_state=42)
y_kmeans = kmeans.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', marker='X')
plt.show()
```

47.Generate noisy blobs using make_blobs and use DBSCAN to identify both clusters and noise points. Print the
count.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data with noise
X, _ = make_blobs(n_samples=500, centers=4, cluster_std=1.5, random_state=42)

# Apply DBSCAN clustering
dbscan = DBSCAN(eps=1.0, min_samples=5)
labels = dbscan.fit_predict(X)

# Count clusters and noise points
num_clusters = len(set(labels)) - (1 if -1 in labels else 0)
num_noise = np.sum(labels == -1)

print(f"Number of clusters found: {num_clusters}")
print(f"Number of noise points: {num_noise}")

# Visualization
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma', marker='o', edgecolors='k')
plt.title("DBSCAN Clustering with Noisy Blobs")
plt.show()
```

48. Load the Digits dataset, reduce dimensions using t-SNE, then apply Agglomerative Clustering and plot the
clusters.

```python
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Load the Digits dataset
digits = load_digits()
X = digits.data

# Reduce dimensions using t-SNE
X_tsne = TSNE(n_components=2, perplexity=30, random_state=42).fit_transform(X)

# Apply Agglomerative Clustering
agg_clustering = AgglomerativeClustering(n_clusters=10)
labels = agg_clustering.fit_predict(X_tsne)

# Visualization
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='rainbow', marker='o', edgecolors='k')
plt.title("Agglomerative Clustering on t-SNE Reduced Digits Data")
plt.show()
```
